```
#######----------Matplotlib-----------------


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```
#bar chart
x=np.array([10,20,30])
y=np.array([40,50,60])
plt.plot(x,y)
plt.show()
```

```
x=np.array([10,20,30])
y=np.array([40,50,60])
plt.plot(x,y,marker="o")
plt.show()
```

```
x=np.array([10,20,30])
y=np.array([40,50,60])
plt.plot(x,y,marker="*")
plt.show()
```

```
x=np.array([10,20,30])
y=np.array([40,50,60])
plt.plot(x,y,marker="o",mec="Red",mfc="orange",ms=20)
plt.show()
```

```
# line style
#linestyle="dashed","dotted","dashdot"
x=np.array([10,20,30])
y=np.array([40,50,60])
plt.plot(x,y,marker="o",mec="Red",mfc="orange",ms=20,linestyle="dashed",linewidth=4)
plt.grid()

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Chart")
plt.show()
```
